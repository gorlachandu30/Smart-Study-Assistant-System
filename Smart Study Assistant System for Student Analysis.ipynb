## Data Cleanings

```python
def remove_duplicates(data):
    before = data.shape[0]
    data = data.drop_duplicates()
    after = data.shape[0]
    print("Duplicates Removed:", before - after)
    return data
df = remove_duplicates(df)
```

```python
def dataset_summary(data):
    print("\nFINAL DATASET INFO")
    print("--------------------")
    print(data.info())
    print("\nStatistical Summary:\n", data.describe())
dataset_summary(df)
```

```python
def convert_datatypes(data):
    data['AGE'] = data['AGE'].astype(int)
    data['STUDY_HOUR'] = data['STUDY_HOUR'].astype(float)
    data['ATTENDANCE_PERCENTAGE'] = data['ATTENDANCE_PERCENTAGE'].astype(float)
    print("Data types updated!")
    return data
df = convert_datatypes(df)
```

```python
df.to_csv("clean_student_data.csv", index=False)
print("Clean dataset saved successfully!")
```

## Data Filtering

```python
# To Filter students with Attendance above 80%..
high_attendance = df[df['ATTENDANCE_PERCENTAGE'] > 80]
print(high_attendance)
```

```python
# To Filter students with Grade A..
grade_a_students = df[df['GRADE'] == 'A']
print(grade_a_students)
```

```python
# To Filter students studying more than 4 hours..
study_hours = df[df['STUDY_HOUR'] > 4]
print(study_hours)
```

```python
# To Filter students whose weak subject is Mathematics..
math_week = df[df['WEEK_SUBJECT'] == 'Mathematics']
print(math_week)
```

## Data Grouping

```python
# To Group by Grade and find average Final Score
grade_group = df.groupby('GRADE')['FINAL_SCORE'].mean()
print(grade_group)
```

```python
# To Group by Performance and calculate average Study Hours
performance_group = df.groupby('PERFORMANCE')['STUDY_HOUR'].mean()
print(performance_group)
```

```python
# To Group by Weak Subject and count students..
subject_group = df.groupby('WEEK_SUBJECT').size()
print(subject_group)
```

```python
# To Group by Grade and calculate average Attendance
attendance_group = df.groupby('GRADE')['ATTENDANCE_PERCENTAGE'].mean()
print(attendance_group)
```

```python
# To Group by Performance and calculate average scores
score_group = df.groupby('PERFORMANCE')[
    ['ASSIGNMENT_SCORE', 'QUIZE_SCORE', 'MIDTERM_SCORE', 'FINAL_SCORE']
].mean()
print(score_group)
```

## Combined Filtering and Grouping

```python
# To Average Final Score of students with Attendance > 80%
filtered_group = (
    df[df['ATTENDANCE_PERCENTAGE'] > 80]
    .groupby('GRADE')['FINAL_SCORE']
    .mean()
)
print(filtered_group)
```

```python
# To Top-performing students (Final Score > 90) grouped by Grade
top_students = (
    df[df['FINAL_SCORE'] > 90]
    .groupby('GRADE')
    .size()
)
print(top_students)
```

## Data Visualisation

```python
plt.figure(figsize=(6,4))
sns.countplot(x='PERFORMANCE', data=df)
plt.title("Student Performance Distribution")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.barplot(x='GRADE', y='FINAL_SCORE', data=df)
plt.title("Grade vs Final Score")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.scatterplot(x='STUDY_HOUR', y='FINAL_SCORE', hue='PERFORMANCE', data=df)
plt.title("Study Hours vs Final Score")
plt.show()
```

```python
weak_counts = df['WEEK_SUBJECT'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(weak_counts, labels=weak_counts.index, autopct='%1.1f%%')
plt.title("Weak Subject Distribution")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.hist(df['STUDY_HOUR'], bins=5, color='skyblue', edgecolor='black')
plt.title("Study Hours Distribution")
plt.xlabel("Study Hours")
plt.ylabel("Number of Students")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='PERFORMANCE', y='ATTENDANCE_PERCENTAGE', data=df)
plt.title("Attendance by Performance")
plt.show()
```

```python
grade_avg = df.groupby('GRADE')['FINAL_SCORE'].mean().reset_index()

plt.figure(figsize=(6,4))
sns.barplot(x='GRADE', y='FINAL_SCORE', data=grade_avg)
plt.title("Average Final Score by Grade")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title("Correlation Between Variables")
plt.show()
```
